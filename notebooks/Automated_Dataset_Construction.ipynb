# Constructing our dataset of Audio-Image pairs
Slight problem! Our MusicCaps dataset does not contain song titles and it is missing a lot of major songs we'd want to include, so we are going to have to do some manual searching with the Deezer API.

```python
import requests
import IPython.display as ipd
from IPython.display import Image, display
```

```python
search_queries = [
    "deadmau5 - Strobe",
    "Daft Punk - One More Time",
    "Avicii - Levels",
    "Darude - Sandstorm",
    "Skrillex - Scary Monsters and Nice Sprites",
    "New Order - Blue Monday",
    "Martin Garrix - Animals",
    "David Guetta - Titanium",
    "Alan Walker - Faded",
    "Eric Prydz - Opus",
    "Clarity - Zedd, Foxes",
    "Don't You Worry Child - Radio Edit - Swedish House Mafia, John Martin",

]
```

```python
import requests
import pandas as pd

curator_name = "Summer"
genre = "Electronic"

spreadsheet_data = []

for search_query in search_queries:
    url = f"https://api.deezer.com/search?q={search_query}&limit=1"
    response = requests.get(url).json()

    if 'data' in response and len(response['data']) > 0:
        track = response['data'][0]

        spreadsheet_data.append({
            "Curator_Name": curator_name,
            "Genre": genre,
            "Track_ID": track['id'],
            "Song_Name": track['title'],
            "Artist": track['artist']['name'],
            "Deezer_Audio_URL": track.get('preview', 'No Preview Available'),
            "Deezer_Image_URL": track['album'].get('cover_medium', 'No Cover Available')
        })
    else:
        print(f"No results found for: {search_query}")

df = pd.DataFrame(spreadsheet_data)

print(df)

df.to_csv("deezer_songs.csv", index=False)
print("\nSaved to deezer_songs.csv.")
```

Now we are going to augment the dataset a bit by adding more tracks for your genre(s). You will need to use the link here and search for your genre id and add it to the code below: [https://api.deezer.com/genre](https://api.deezer.com/genre)

```python
import requests
import pandas as pd
```

```python
GENRE_NAME = "Pop" # <-- change this
GENRE_ID = 132 # <-- also this

url = f"https://api.deezer.com/chart/{GENRE_ID}/tracks?limit=100"
response = requests.get(url).json()

valid_tracks = []

if 'data' in response:
    for track in response['data']:
        preview_url = track.get('preview')
        image_url = track['album'].get('cover_xl')

        if preview_url and image_url:
            valid_tracks.append({
                "Curator_Name": "Alex", # <-- change to YOUR NAME
                "Genre": GENRE_NAME,
                "Track_ID": track['id'],
                "Song_Name": track['title'],
                "Artist": track['artist']['name'],
                "Deezer_Audio_URL": preview_url,
                "Deezer_Image_URL": image_url
            })

filename = f"{GENRE_NAME}_expanded_data.csv"
df = pd.DataFrame(valid_tracks)
df.to_csv(filename, index=False)
```
